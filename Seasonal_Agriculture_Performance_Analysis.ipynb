```python
# VOIS AICTE Batch 2026-2027 | Major Project: Seasonal Agriculture Performance Analysis
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configure visual styling
sns.set_theme(style="whitegrid")
%matplotlib inline
```

```python
# Objective: Explore and understand the dataset structure, features, and types
df = pd.read_csv('seasonal_agriculture_performance_dataset (1).csv')

print("--- 1. DATASET EXPLORATION ---")
print(f"Dataset Shape (Rows, Columns): {df.shape}")
print("\nColumn Data Types & Non-Null Counts:")
df.info()

print("\nFirst 5 Records:")
display(df.head())

print("\nSummary Statistics of Numerical Features:")
display(df.describe())
```

```python
# Objective: Clean and prepare the data for analysis (Handling missing values)
print("--- 2. DATA CLEANING & PREPARATION ---")
print("Missing values before cleaning:\n", df.isnull().sum())

# Contextual Imputations:
# 1. Rainfall & Soil Moisture: Imputed using median grouped by State and Season
df['Rainfall_mm'] = df['Rainfall_mm'].fillna(df.groupby(['State', 'Season'])['Rainfall_mm'].transform('median'))
df['Soil_Moisture_pct'] = df['Soil_Moisture_pct'].fillna(df.groupby(['State', 'Season'])['Soil_Moisture_pct'].transform('median'))

# 2. Yield: Imputed using median grouped by Crop and Season
df['Yield_Tonnes_Ha'] = df['Yield_Tonnes_Ha'].fillna(df.groupby(['Crop', 'Season'])['Yield_Tonnes_Ha'].transform('median'))

print("\nMissing values after cleaning:\n", df.isnull().sum())
```

```python
# Key Questions Addressed:
# - How does agricultural performance vary across seasons?
# - What major seasonal patterns can be observed?
print("--- 3. SEASONAL PERFORMANCE & VARIATIONS ---")

seasonal_performance = df.groupby('Season').agg({
    'Yield_Tonnes_Ha': ['mean', 'median', 'std'],
    'Production_Tonnes': ['mean', 'sum'],
    'Profit_INR': ['mean', 'median']
}).reset_index()

display(seasonal_performance)

# Visualization: Yield and Profit Variations Across Seasons
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.barplot(data=df, x='Season', y='Yield_Tonnes_Ha', ax=axes[0], palette='viridis', errorbar=None)
axes[0].set_title('Average Yield by Season (Tonnes/Ha)')
axes[0].set_ylabel('Yield (Tonnes/Ha)')

sns.barplot(data=df, x='Season', y='Profit_INR', ax=axes[1], palette='magma', errorbar=None)
axes[1].set_title('Average Profit by Season (INR)')
axes[1].set_ylabel('Profit (INR)')
plt.tight_layout()
plt.show()
```

```python
# Key Questions Addressed:
# - Which characteristics change between seasons?
# - Are there noticeable variations in resource usage across seasons?
# - Are there relationships between seasonal environmental conditions and agricultural performance?
print("--- 4. ENVIRONMENTAL & RESOURCE USAGE ANALYSIS ---")

resource_env_summary = df.groupby('Season').agg({
    'Rainfall_mm': 'mean',
    'Avg_Temperature_C': 'mean',
    'Humidity_pct': 'mean',
    'Soil_Moisture_pct': 'mean',
    'Water_Used_m3': 'mean',
    'Water_Efficiency_t_per_1000m3': 'mean',
    'Fertilizer_kg_ha': 'mean',
    'Pesticide_Litre_ha': 'mean'
}).reset_index()

display(resource_env_summary)

# Visualization: Environmental Conditions Across Seasons
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.boxplot(data=df, x='Season', y='Rainfall_mm', ax=axes[0], palette='Set2')
axes[0].set_title('Rainfall Distribution (mm) Across Seasons')

sns.boxplot(data=df, x='Season', y='Avg_Temperature_C', ax=axes[1], palette='Set2')
axes[1].set_title('Average Temperature (°C) Across Seasons')
plt.tight_layout()
plt.show()
```

```python
# Key Questions Addressed:
# - How do economic outcomes vary across seasons?
# - Are some seasonal patterns consistent across different regions or categories?
print("--- 5. ECONOMIC OUTCOMES & REGIONAL CONSISTENCY ---")

economic_summary = df.groupby('Season').agg({
    'Total_Cost_INR': 'mean',
    'Revenue_INR': 'mean',
    'Profit_INR': 'mean'
}).reset_index()
display(economic_summary)

regional_profit = df.groupby(['State', 'Season'])['Profit_INR'].mean().unstack()
print("\nAverage Profit (INR) Across States and Seasons (Regional Consistency):")
display(regional_profit)

# Visualization: Regional Profit Consistency
plt.figure(figsize=(12, 6))
sns.barplot(data=df, x='State', y='Profit_INR', hue='Season', palette='coolwarm', errorbar=None)
plt.title('Regional Profitability Consistency Across States & Seasons')
plt.xticks(rotation=30)
plt.ylabel('Average Profit (INR)')
plt.legend(title='Season')
plt.show()
```

```python
# Key Questions Addressed:
# - Are there unusual or unexpected seasonal patterns?
# - What differences exist between agricultural activities in different seasons?
print("--- 6. STATISTICAL CORRELATIONS & UNUSUAL PATTERNS ---")

numeric_cols = ['Rainfall_mm', 'Avg_Temperature_C', 'Humidity_pct', 'Soil_Moisture_pct',
                'Fertilizer_kg_ha', 'Seed_Quality_Score', 'Yield_Tonnes_Ha', 'Profit_INR', 'Disease_Pest_Risk_pct']

plt.figure(figsize=(10, 8))
correlation_matrix = df[numeric_cols].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='Blues', fmt='.2f', linewidths=0.5)
plt.title('Correlation Matrix: Environmental & Agricultural Outcomes')
plt.show()

print("\nUnusual Pattern Check - Zaid Season Economic Strains:")
print(df.groupby('Season')[['Profit_INR', 'Water_Used_m3']].mean())
```
